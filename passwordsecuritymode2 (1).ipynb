```python
# PASSWORD SECURITY SYSTEM

EXIT_PASSWORD = "admin_exit"

while True:
    # Get password from user
    password = input(f"Enter your password (or type the exit password '{EXIT_PASSWORD}' to quit): ")

    if password == EXIT_PASSWORD:
        print("Exiting password registration system.")
        break

    # Check password strength
    if len(password) < 6:
        strength = "LOW"
    elif (len(password) >= 8 and
          any(c.isupper() for c in password) and
          any(c.islower() for c in password) and
          any(c.isdigit() for c in password) and
          any(not c.isalnum() for c in password)):
        strength = "HIGH"
    else:
        strength = "MEDIUM"

    print("Password Strength:", strength)

    # Read existing passwords from file
    try:
        file = open("password.txt", "r")
        passwords = file.read().splitlines()
        file.close()
    except FileNotFoundError:
        passwords = []

    # Check duplicate password
    if password in passwords:
        print("Duplicate Password Found!")
        print("Registration Failed!\n")
    else:
        # Save new password in file
        file = open("password.txt", "a")
        file.write(password + "\n")
        file.close()
        print("Password Registered Successfully!\n")
```
